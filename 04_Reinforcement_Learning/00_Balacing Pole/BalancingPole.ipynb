<h1>Reinforcement Learning</h1>

Reinforcement Learning (Verstärkendes Lernen) ist ein Teilgebiet des maschinellen Lernens, bei dem ein sogenannter Agent durch Versuch und Irrtum lernt, Entscheidungen zu treffen. Der Agent interagiert mit seiner Umgebung, führt Aktionen aus und erhält dafür Belohnungen oder Bestrafungen. Ziel ist es, eine Strategie zu entwickeln, die langfristig die größtmögliche Belohnung erzielt. Im Gegensatz zum überwachten Lernen werden dem System keine korrekten Lösungen vorgegeben. Stattdessen lernt es selbstständig aus den Folgen seiner Handlungen. Reinforcement Learning wird unter anderem in der Robotik, bei autonomen Fahrzeugen, in der Spieleentwicklung und zur Optimierung komplexer Prozesse eingesetzt.

<h2>Balanzieren eines Stabes (Balancing-Pole-Problem)</h2>

Das Balancing-Pole-Problem ist ein klassisches RL-Testproblem, bei dem ein Agent lernen soll, einen Stab (Pole), der auf einem Wagen (Cart) montiert ist, aufrecht zu halten.

Der Wagen kann sich nur nach links oder rechts bewegen.
Die einzige Aktion des Agenten ist also:

- Wagen nach links schieben

- Wagen nach rechts schieben

<h3>Zustand (State)</h3>

Der Agent bekommt pro Zeitschritt typischerweise diese Werte:

- Position des Wagens

- Geschwindigkeit des Wagens

- Winkel des Stabs

Winkelgeschwindigkeit des Stabs

<h3>Ziel</h3>

Der Stab soll so lange wie möglich aufrecht bleiben. Jeder Zeitschritt, in dem der Stab nicht umfällt, gibt eine Belohnung von +1. Die Episode endet, wenn:

- der Stab zu stark kippt (z. B. > 12°), oder

- der Wagen aus dem Bereich fährt (z. B. x > 2.4)

<h2>Gymnasium</h2>

Gymnasium ist eine Python-Bibliothek von Open AI für Reinforcement Learning. Sie stellt standardisierte Simulationsumgebungen bereit, in denen RL-Agenten trainiert und getestet werden können. Gymnasium stellt eine einheitliche Schnittstelle für viele verschiedene Aufgaben bereit. Dadurch kann derselbe RL-Algorithmus auf unterschiedlichen Problemen verwendet werden.

<h3>Grundidee</h3>

Ein Agent kommuniziert mit einer Umgebung:

- Agent  ---> Aktion ----> Umgebung
- Agent <--- Zustand ----- Umgebung
- Agent <--- Belohnung --- Umgebung

Der Agent

- beobachtet den aktuellen Zustand,
- wählt eine Aktion,
- erhält den neuen Zustand und eine Belohnung,
- lernt daraus eine bessere Strategie.

<h2>Das Grundgerüst</h2>

Das folgende Programm baut eine Gymnsium Umgebung für das Balazierproblem auf und führt 10 Durchläufe aus:

```python
import gymnasium as gym

# Trainingsummgebung für Stabbalanzieren
env = gym.make("CartPole-v1", render_mode="human")

for episode in range(1, 11):
    score = 0

     # Startet eine neue Episode
    state = env.reset()
    terminated = False

    while not terminated:
        env.render()

        # action_space.sample() wählt zufällig: 0 = nach links schieben  1 = nach rechts schieben

        action = env.action_space.sample()


        observation, reward, terminated, truncated, info = env.step(action)
        
        # von Step() erhält man:

        # observation: aktueller Zustand
        # reward: Belohnung (1 pro Zeitschritt)
        # terminated: Aufgabe erfolgreich beendet oder Stab umgefallen
        # truncated: Zeitlimit erreicht
        # info: zusätzliche Informationen

        # Der Zustandsvektor observation besteht aus vier Werten:

        #  Index	Bedeutung
        #  0	Position des Wagens
        #  1	Geschwindigkeit des Wagens
        #  2	Winkel des Stabs
        #  3	Winkelgeschwindigkeit des Stabs

        score += reward

    print('Episode:', episode, 'Score:', score, 'State:' , observation)
env.close()
```

<h2>Ergebnis</h2>

Was wurde also bisher gelernt? Gar nichts! Das gezeigte Beispiel enthält überhaupt keinen Lernalgorithmus.

Die Zeile

action = env.action_space.sample()

wählt jedes Mal eine zufällige Aktion. Der Agent speichert keine Erfahrungen, passt keine Parameter an und verbessert sich nicht. Das Beispiel demonstriert lediglich, wie man mit einer Gymnasium-Umgebung interagiert.

Der Ablauf ist:

- Zustand lesen
- Zufällige Aktion wählen
- Aktion ausführen
- Belohnung erhalten
- Wiederholen

Das ist eher ein Test, ob die Umgebung funktioniert.

Damit tatsächlich gelernt wird, benötigt man zusätzlich einen Reinforcement-Learning-Algorithmus, beispielsweise:

- Q-Learning
- SARSA
- Deep Q-Network (DQN)
- PPO
- ...
  
Dann sieht der Ablauf so aus:

- Zustand beobachten
        
- Aktion auswählen
        
- Belohnung erhalten
        
- Erfahrung speichern
        
- Strategie aktualisieren
        
- Nächster Zustand

Zum Beispiel könnte ein DQN-Agent lernen:

- Wenn der Stab nach rechts kippt → Wagen nach rechts bewegen.
- Wenn der Stab nach links kippt → Wagen nach links bewegen.
- Je besser der Stab balanciert wird → desto mehr Belohnung erhält der Agent.

Nach vielen Episoden entwickelt der Agent dann eine Strategie, die den Stab oft die maximalen 500 Schritte aufrecht hält.

<h2>Implementierung einer Lernstrategie</h2>

Es gibt viele Implementierungen von Lernstrategie für Reinforcement Learning. Es gibt sowohl klassische Ansätze (Q-Learning, Entscheidungsbäume, ..) als auch Lösungen, die Neuronale Netze verwenden.

Als Beispiel für dieses Tutorial dient PPO: PPO (Proximal Policy Optimization) gehört heute zu den am häufigsten verwendeten Reinforcement-Learning-Algorithmen, weil es vergleichsweise stabil trainiert und mit kontinuierlichen sowie diskreten Aktionen umgehen kann.

Stable-Baselines3 (SB3) ist eine Python-Bibliothek, die fertige Implementierungen vieler Reinforcement-Learning-Algorithmen bereitstellt, darunter auch PPO.

<h2>Training</h2>

MlpPolicy ist in Stable-Baselines3 die Standard-Policy für Zustände, die als Zahlenvektoren vorliegen. MLP steht für Multi-Layer Perceptron, also ein gewöhnliches Feedforward Netz.

model = PPO("MlpPolicy", env)

bedeutet:
- PPO ist der Lernalgorithmus.
- MlpPolicy ist die Art des neuronalen Netzes, das PPO verwendet.


```python
import gymnasium as gym
from stable_baselines3 import PPO

# Trainingsummgebung für Stabbalanzieren
env = gym.make("CartPole-v1")

model = PPO('MlpPolicy', env, verbose=1)
model.learn(total_timesteps=20000)

# save the model
model.save('ppo balancing model')
```

<h2>Evaluation der Ergebnisse</h2>

Nun lassen wir den Stab noch mal 10 Durchgänge lang balanzieren, diesmal aber natürlich nicht mit einer zufälligen Aktion wie am Anfang sondern mit der Aktion, die vom trainierten Netz vorgeschlagen wird. Man beachte die Unterschiede im Score zum Eingangsbeispiel!

```python
import gymnasium as gym
from stable_baselines3 import PPO

env = gym.make("CartPole-v1", render_mode="human")

model = PPO.load("ppo balancing model", env=env)

for episode in range(1, 11):
    score = 0
    obs, info = env.reset()
    done = False

    while not done:
        env.render()
        
        # die Aktion wird nun durch das trainierte Modell festgelegt
        
        action, _ = model.predict(obs)
        obs, reward, done, info, whatever = env.step(action)
        score += reward

    print('Episode:', episode, 'Score:', score)
env.close()
```
