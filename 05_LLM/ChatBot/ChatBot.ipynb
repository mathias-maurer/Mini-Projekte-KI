## 3. Die Python-Klasse gpt4all.GPT4All
### Konstruktor: \_\_init\_\_(self, model_name)
<pre>
Erstellt und Initialisiert ein GPT4All-Objekt

Argumente:
  self:                 -- aktuelles GPT4All-Objekt
  model_name: str       -- Name oder vollständiger Pfad der Modell-Datei als String
  allow_download: bool  -- Internetzugriff erlauben
                              True  -> Es wird im Internet nach der Modelldatei gesucht, auch wenn bereits vorhanden!!
                              False -> Modelldatei wird nur lokal gesucht.
                              (Default: True)
</pre>
### Methode: generate(self, prompt, max_tokens, temp, streaming)
<pre>
Generiert passenden Text zu einem Prompt.

Argumente:
  self:           -- aktuelles GPT4All-Objekt
  prompt: str     -- Benutzereingabe bzw. Prompt
  max_tokens: int -- maximale Anzahl der Ausgabe-Tokens 
                     (Default: 200)
  temp: float     -- Temperaturwert: 
                       0  -> Antworten bei gleichem Prompt immer gleich,
                       >0 -> Antworten variaeren bei gleichem Prompt
                       (Default: 0.7)
  streaming: bool -- Streamingmodus aktivieren
                       True  -> Es wird ein Generator-Objekt zurückgegeben, aus dem die Token einzeln entnommen werden können.
                       False -> Die Antwort wird in einem einzigen Gesamt-String zurückgegeben.
                       (Default: False)

Rückgabe:
  str        -- Antwort als Gesamt-String
  Generator  -- wenn Streamingmodus aktiv
</pre>

## 5. System-Prompt ändern
Der System-Prommpt ist ein Text, der das Verhalten bzw. die Persona des Sprachmodells steuert und bei jeder Anfrage vor den eigentlichen Benutzer-Prompt gesetzt wird.

Damit das Sprachmodell System-Prompt und aktuelle Benutzerfrage unterscheiden kann, sind einige Spezial-Token eintrainiert worden.

Bei Llama-Modellen sind dies unter anderem:
<pre>
  <|begin_of_text|>     -- Beginn des Gesamt-Prompts
  <|start_header_id|>   -- Beginn der Kopfzeile eines Prompt-Abschnitts
  <|end_header_id|>     -- Ende der Kopfzeile eines Prompt-Abschnitts
  <|eot_id|>            -- Ende eines Prompt-Abschnitts
</pre>

Hier ein Beispiel für einen Gesamt-Prompt:
<pre>
  <|begin_of_text|>
  <|start_header_id|>system<|end_header_id|>
  Du bist ein hilfreicher Assistent. Gib alle Antworten auf Deutsch!
  <|eot_id|>

  <|start_header_id|>user<|end_header_id|>
  Was ist die Hauptstadt von Frankreich?
  <|eot_id|>

  <|start_header_id|>assistant<|end_header_id|>
</pre>
