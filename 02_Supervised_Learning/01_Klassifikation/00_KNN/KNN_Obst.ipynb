# Beispiel Obstklassifikation

In diesem Notebook demonstrieren wir KNN (k-Nearest Neighbors) anhand eines Obstbeispiels.
Wir klassifizieren neue Obstproben (Äpfel und Orangen) anhand von Größe und Farbe.

```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt
```

Hier werden die Daten definiert:

```python
data = {
    'Größe': [7, 6, 8, 10, 9, 11],
    'Farbe': [3, 4, 2, 8, 7, 9],
    'Klasse': ['Apfel', 'Apfel', 'Apfel', 'Orange', 'Orange', 'Orange']
}

df = pd.DataFrame(data)

X_train = df[['Größe', 'Farbe']].values
y_train = df['Klasse'].values

# --- Neue Datenpunkte ---
X_new = np.array([[8, 4], [9, 6]])
```

Anschließend visualisiert

```python
# Trainingsdaten plotten
for klasse, marker, color in zip(['Apfel', 'Orange'], ['o', 's'], ['red', 'orange']):
    plt.scatter(df[df['Klasse']==klasse]['Größe'], df[df['Klasse']==klasse]['Farbe'], 
                marker=marker, color=color, s=100, label=klasse)

# Neue Datenpunkte plotten
plt.scatter(X_new[:,0], X_new[:,1], marker='*', color='blue', s=200, label='Neue Daten')

plt.xlabel('Größe (cm)')
plt.ylabel('Farbe (1=grün, 10=rot)')
plt.title('KNN Klassifikation von Obst')
plt.legend()
plt.grid(True)
plt.show()
```

Hier wird das Modell erstellt:

```python
k = 3
knn = KNeighborsClassifier(n_neighbors=k)
knn.fit(X_train, y_train)
```

Und jetzt die Vorhersage getroffen:

```python
y_pred = knn.predict(X_new)

# --- Ausgabe ---
for i, point in enumerate(X_new):
    print(f"Punkt {point} wird klassifiziert als: {y_pred[i]}")
```
