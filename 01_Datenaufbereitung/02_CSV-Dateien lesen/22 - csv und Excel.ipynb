# pandas CSV- und Excel-Ein-/Ausgabe
Pandas ist eine leistungsfähige Python-Bibliothek zur Datenanalyse und -aufbereitung. Sie bietet zwei zentrale Datenstrukturen:
- Series: eindimensionale, beschriftete Daten (vergleichbar mit einer Spalte in Excel).
- DataFrame: zweidimensionale, tabellarische Datenstruktur mit Zeilen und Spalten.

Mit Pandas können Sie große Datensätze effizient bearbeiten, analysieren und vorbereiten


## `pd.read_csv()` -- CSV-Dateien einlesen

`pandas.read_csv()` lädt Tabellen aus CSV-Dateien in ein DataFrame.

### Grundsyntax

``` python
import pandas as pd
df = pd.read_csv("datei.csv")
```

### Wichtige Parameter

-   **sep** -- Trennzeichen
-   **decimal** -- Dezimaltrennung


### Beispiele

``` python
df = pd.read_csv("data.csv", sep=",", decimal=".")
df = pd.read_csv("data.csv", sep=";", decimal=",")
df = pd.read_csv("data.csv", sep="\t")

```

------------------------------------------------------------------------

## CSV-Dateien schreiben -- `to_csv()`

### Grundsyntax

``` python
df.to_csv("export.csv")
```

### Wichtige Parameter

-   **sep** -- Trennzeichen
-   **decimal** -- Dezimalzeichen
-   **index** -- Index speichern?


### Beispiele

``` python
df.to_csv("daten.csv", index=False)
df.to_csv("daten.csv", sep=";", decimal=",", index=False)
```

------------------------------------------------------------------------

## Excel-Dateien exportieren -- `to_excel()`

### Grundsyntax

``` python
df.to_excel("daten.xlsx")
```


### Beispiele

``` python
df.to_excel("daten.xlsx")
df.to_excel("daten.xlsx", index=False)
```

------------------------------------------------------------------------

## Zusammenfassung

-   **CSV lesen**: `pd.read_csv()`
-   **CSV schreiben**: `df.to_csv()`
-   **Excel schreiben**: `df.to_excel()`
